```python
import pandas as pd
import numpy as np
# Load dataset from the CSV file
df = pd.read_csv('ecommerce_customer_data.csv')
print(" Dataset Head ")
print(df.head())
```

```python

# 1. Total Orders and Completed-Order Value

total_orders = len(df)
completed_orders = df[df['order_status'] == 'Completed']
total_completed_value = completed_orders['order_value'].sum()
print("\n# 1. Order Summary :")
print(f"Total Orders: {total_orders}")
print(f"Total Completed Order Value: ₹{total_completed_value:,}")
```

```python

# 2. Categories and Cities Contributing Most Orders & Sales

cat_summary = df.groupby('category')['order_value'].agg(['count', 'sum']).rename(columns={'count': 'orders', 'sum': 'total_sales'}).sort_values(by='total_sales', ascending=False)
city_summary = df.groupby('city')['order_value'].agg(['count', 'sum']).rename(columns={'count': 'orders', 'sum': 'total_sales'}).sort_values(by='total_sales', ascending=False)
print("\n# 2. Top Categories :")
print(cat_summary)
print("\n# 2. Top Cities :")
print(city_summary)
```

```python

# 3. Proportion of Customers Placing More than One Order

order_counts = df.groupby('customer_id')['order_id'].count()
repeat_customers = (order_counts > 1).sum()
total_customers = df['customer_id'].nunique()
repeat_proportion = (repeat_customers / total_customers) * 100
print("\n# 3. Repeat Customer Proportion :")
print(f"Repeat Customers: {repeat_customers} out of {total_customers} ({repeat_proportion:.2f}%)")
```

```python

# 4. Average Order Value (AOV) by Customer Groups

aov_by_group = df.groupby('customer_segment')['order_value'].mean().round(2)
print("\n# 4. Average Order Value by Segment :")
print(aov_by_group)
```

```python

# 5. Rates of Cancellation and Return
status_pct = (df['order_status'].value_counts() / len(df)) * 100
print("\n# 5. Order Status Percentages :)
print(f"Cancellation Rate: {status_pct.get('Cancelled', 0):.2f}%")
print(f"Return Rate: {status_pct.get('Returned', 0):.2f}%")
```

```python

# 6. Delivery Duration vs Customer Rating

avg_rating_by_days = df.groupby('delivery_days')['customer_rating'].mean().round(2)
corr = df['delivery_days'].corr(df['customer_rating'])
print("\n# 6. Delivery Time vs Ratings :")
print(avg_rating_by_days)
print(f"Correlation: {corr:.2f} (Negative correlation confirms delays reduce ratings)")
```

```python

# 7. Customer Segments Showing Stronger Repeat Purchase

cust_info = df.groupby(['customer_id', 'customer_segment'])['order_id'].count().reset_index()
cust_info['is_repeat'] = cust_info['order_id'] > 1
repeat_by_segment = (cust_info.groupby('customer_segment')['is_repeat'].mean() * 100).round(2)
print("\n# 7. Repeat Purchase % by Segment ")
print(repeat_by_segment.astype(str) + '%')
```
